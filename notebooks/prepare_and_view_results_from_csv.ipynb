```python
import sys
sys.path.append("../")

from odometry.analyzers.analyzer import Analyzer

#specify the save folder
# save_folder = "/home/david/Documents/odometry/scripts/NaiveRadar09112024/Results"
save_folder = "/home/david/Documents/odometry/scripts/OdomOnly09112024/Results"
# save_folder = "/home/david/Documents/odometry/scripts/Radnav09112024/Results"
# save_folder ="/home/david/Documents/odometry/scripts/radarHD09112024/Results"
# save_folder ="/home/david/Documents/odometry/scripts/radCloud09112024/Results"


#initialize the analyzer
analyzer = Analyzer()

analyzer.show_cumulative_summary_from_csvs(
    save_folder=save_folder
)
```

```python
from odometry.plotting.plotter_analyzer import PlotterAnalyzer
import matplotlib.pyplot as plt
plotter = PlotterAnalyzer()

#get absolute position errors
absolute_errors_pos,absolute_errors_heading = \
    analyzer.get_absolute_errors_from_csvs(save_folder)

fig,ax = plt.subplots(figsize=(5,1.2))

plotter.plot_error_histogram(
    errors=absolute_errors_pos,
    bins=50,
    ax=ax,
    show=False
)
ax.set_title("")
ax.set_xlabel("ATE Error (m)")
```

```python
from odometry.plotting.plotter_analyzer import PlotterAnalyzer
import matplotlib.pyplot as plt
plotter = PlotterAnalyzer()

#get absolute position errors
relative_errors_pos,relative_errors_heading = \
    analyzer.get_relative_errors_from_csvs(save_folder)

fig,ax = plt.subplots(figsize=(10,1.2))

plotter.plot_error_histogram(
    errors=relative_errors_pos,
    bins=100,
    ax=ax,
    show=False
)
ax.set_title("")
ax.set_xlabel("RTE Error (m)")
```
